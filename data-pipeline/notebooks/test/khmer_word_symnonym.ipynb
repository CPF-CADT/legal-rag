```python
from pathlib import Path

PROJECT_ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / 'react-client').is_dir() and (path / 'data-pipeline').is_dir()
)
DATA_DIR = PROJECT_ROOT / 'data-pipeline' / 'data'
TESTABLE_DIR = PROJECT_ROOT / 'data-pipeline' / 'data' / 'csv'

import pandas as pd

df = pd.read_parquet(DATA_DIR / 'khmer-lexicon' / 'words.parquet')

df.to_csv(TESTABLE_DIR / 'experimental_words.csv', index=False, encoding='utf-8-sig')

print(f"Exported {len(df)} rows")
```
