# Shared embedding pipeline

This notebook documents the common model configuration, input format, output format, normalization, and validation used by the article-level and chunk-level embedding notebooks.

The generated vectors are L2-normalized and their row order matches the metadata JSONL file.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'data-pipeline').is_dir() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_DIR = PROJECT_ROOT / 'data-pipeline' / 'data'
SEARCH_DIR = DATA_DIR / 'processed' / 'search'
EMBEDDING_DIR = DATA_DIR / 'processed' / 'embeddings'

MODELS = {
    'sealion': {
        'model': 'aisingapore/SEA-LION-E5-Embedding-600M',
        'provider': 'sentence-transformer',
        'batch_size': 8,
        'output': 'sealion_e5_600m',
    },
    'granite-r2': {
        'model': 'ibm-granite/granite-embedding-311m-multilingual-r2',
        'provider': 'sentence-transformer',
        'batch_size': 2,
        'output': 'granite_311m_r2',
    },
    'labse': {
        'model': 'sentence-transformers/LaBSE',
        'provider': 'sentence-transformer',
        'batch_size': 16,
        'output': 'labse',
    },
    'albert': {
        'model': 'seanghay/albert-khmer-small',
        'provider': 'huggingface-mean-pooling',
        'batch_size': 8,
        'output': 'albert_khmer_small',
    },
}

print('Embedding configuration loaded for:', ', '.join(MODELS))
```

## Output contract

Each model writes three files under `data/processed/embeddings/`: a NumPy vector file, an aligned metadata JSONL file, and a manifest containing the model, input path, row count, and vector dimension.
